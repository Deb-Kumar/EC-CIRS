# 09 - Unified Model Evaluation & Synthesis
## E-Commerce Customer Intelligence System
**Objective:** Consolidate evaluation metrics across all 4 machine learning components (Regression, Classification, Clustering, and Recommendations) and compare against business benchmarks.


```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt

meta_path = os.path.join("..", "models", "metadata.json")
with open(meta_path, 'r') as f:
    meta = json.load(f)
```

### 1. Spending Prediction Performance (Linear Regression)


```python
spend_df = pd.DataFrame([meta['spending_prediction']['metrics']])
spend_df.index = ['Linear Regression']
spend_df
```

### 2. Purchase Prediction Performance (Logistic Regression)


```python
purch_df = pd.DataFrame([meta['purchase_prediction']['metrics']])
purch_df[['accuracy', 'precision', 'recall', 'f1', 'roc_auc']]
```

### 3. Customer Segmentation Performance (K-Means)


```python
k_df = pd.DataFrame.from_dict(meta['customer_segmentation']['k_evaluations'], orient='index')
k_df[['k', 'inertia', 'silhouette_score', 'davies_bouldin_index']]
```

### 4. Segment Centroid Distribution


```python
cluster_df = pd.DataFrame(meta['customer_segmentation']['cluster_profiles'])
cluster_df[['cluster', 'segment_name', 'customer_count', 'avg_revenue', 'avg_purchases']]
```

### 5. Recommendation System Verification


```python
rec_df = pd.DataFrame(meta['recommendation_system']['sample_recommendation']['recommended'])
rec_df
```
